```python
import pandas as pd
df = pd.read_csv("outputs/bab4_evidence/tables/tab_411_inventaris_workbook.csv")
df[["alias_file","sheet_count","filled_cells","formula_cells","merged_ranges"]]
```

```python
total_sheet = int(df["sheet_count"].sum())
total_filled = int(df["filled_cells"].sum())
total_formula = int(df["formula_cells"].sum())
total_merged = int(df["merged_ranges"].sum())
pct = total_formula / total_filled * 100
print(f"lembar: {total_sheet}")  # 47
print(f"sel terisi: {total_filled}")  # 20050
print(f"formula: {total_formula}")  # 5377
print(f"% formula: {pct:.2f}%")  # 26.82%
print(f"merged: {total_merged}")  # 545 versi commit, 547 working-tree baru (Nomenklatur 52->54)
assert total_sheet == 47
assert total_filled == 20050
assert total_formula == 5377
assert total_merged in (545, 547), f"merged={total_merged}, cek versi Nomenklatur.xlsx"
print("Catatan: 545 = versi commit (Nomenklatur merged=52); 547 = working-tree 2026-09-25 (Nomenklatur merged=54)")
```
